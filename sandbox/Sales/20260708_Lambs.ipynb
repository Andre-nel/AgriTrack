```python
dohne_price = 60.25
mm_price = 53

dohne_weights = [
    40, 46, 30, 29, 54, 37, 36, 39, 38, 25,
    35, 35, 39, 35, 37, 37, 36, 32, 29, 33,
    52, 31, 33, 33, 28, 31, 27, 28, 32, 33,
    34, 35]

mm_weights = [39, 47]

dohne_total_kg = sum(dohne_weights)
dohne_price_total = dohne_total_kg * dohne_price * 1.15

mm_total_kg = sum(mm_weights)
mm_price_total = mm_total_kg * mm_price * 1.15

print(f"Dohne total weight: {dohne_total_kg} kg, average: {dohne_total_kg / len(dohne_weights) if dohne_weights else 0:.2f} kg, total price: R{dohne_price_total:.2f}, average lamb price : R{dohne_price_total / len(dohne_weights) if dohne_weights else 0:.2f}")
print(f"MM total weight: {mm_total_kg} kg, average: {mm_total_kg / len(mm_weights) if mm_weights else 0:.2f} kg, total price: R{mm_price_total:.2f}, average lamb price: R{mm_price_total / len(mm_weights) if mm_weights else 0:.2f}")
print(f'Total money: R{dohne_price_total + mm_price_total:.2f}')
```
